```python
import numpy as np
import scipy as sc
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import time

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.pipeline import Pipeline
import warnings
warnings.filterwarnings('ignore')
```

# Model Training and Cross Validation

```python
#split into x and y
X = df.drop(57, axis=1)
Y = df[57]
```

```python
#define the three models
LR_model = LogisticRegression(max_iter=200)
RF_model = RandomForestClassifier()
SVM_model = SVC()
```

```python
#LOGISTIC REGRESSION CROSS-VALIDATION

#LR pipeline
LR_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("LR_model", LR_model)
])

LR_trainingTime = []
LR_accuracy = []
LR_f1Scores = []

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=7)

for train_index, test_index in skf.split(X, Y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = Y.iloc[train_index], Y.iloc[test_index]
    
    start_time = time.time()
    LR_pipeline.fit(X_train, y_train)
    end_time = time.time()
    
    LR_trainingTime.append(end_time - start_time)
    
    y_pred = LR_pipeline.predict(X_test)
    LR_accuracy.append(accuracy_score(y_test, y_pred))
    LR_f1Scores.append(f1_score(y_test, y_pred, average='macro'))

print('\nLogistic Regression')
print("Training times:", [round(x, 3) for x in LR_trainingTime])
print("Accuracy scores:", [round(x, 3) for x in LR_accuracy])
print("F1 scores:", [round(x, 3) for x in LR_f1Scores])
```

```python
#RANDOM FOREST CLASSIFIER CROSS-VALIDATION
RF_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("RF_model", RF_model)
])

RF_trainingTime = []
RF_accuracy = []
RF_f1Scores = []

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=123)

for train_index, test_index in skf.split(X, Y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = Y.iloc[train_index], Y.iloc[test_index]
    
    start_time = time.time()
    RF_pipeline.fit(X_train, y_train)
    end_time = time.time()
    
    RF_trainingTime.append(end_time - start_time)
    
    y_pred = RF_pipeline.predict(X_test)
    RF_accuracy.append(accuracy_score(y_test, y_pred))
    RF_f1Scores.append(f1_score(y_test, y_pred, average='macro'))

print('\nRandom Forest')
print("Training times:", [round(x, 3) for x in RF_trainingTime])
print("Accuracy scores:", [round(x, 3) for x in RF_accuracy])
print("F1 scores:", [round(x, 3) for x in RF_f1Scores])
```

```python
#SUPPORT VECTOR CLASSIFIER CROSS-VALIDATION
svc_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("SVM_model", SVM_model)
])

svc_trainingTime = []
svc_accuracy = []
svc_f1Scores = []

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=123)

for train_index, test_index in skf.split(X, Y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = Y.iloc[train_index], Y.iloc[test_index]
    
    start_time = time.time()
    svc_pipeline.fit(X_train, y_train)
    end_time = time.time()
    
    svc_trainingTime.append(end_time - start_time)
    
    y_pred = svc_pipeline.predict(X_test)
    svc_accuracy.append(accuracy_score(y_test, y_pred))
    svc_f1Scores.append(f1_score(y_test, y_pred, average='macro'))

print('\nSupport Vector Machine')
print("Training times:", [round(x, 3) for x in svc_trainingTime])
print("Accuracy scores:", [round(x, 3) for x in svc_accuracy])
print("F1 scores:", [round(x, 3) for x in svc_f1Scores])
```

# Friedman Test

```python
#So the three tables are: F1Scores_df, trainingTime_df and accuracy_df - The tables shown above are the stat tables
#Now, I create rank tables to begin the friedman test
#Remember Best (1) to Worst(3). For F1score and accuracy, the higher the better. For Time - the lower the better 

model_cols = ['LR', 'RF', 'SVM'] #apparently i cant just pass the exact values into the next line, I must use this variable
```

### Creating the rank tables for each metric

```python
#Accuracy Rank Table
accuracy_ranks = accuracy_df[model_cols].rank(axis=1, ascending=False)
accuracy_ranks.insert(0, 'Fold', range(1,11)) #adding the fold column
Accuracy_rankMean = accuracy_ranks.mean().rename('Mean')
Accuracy_rankMean['Fold'] = '_'
accuracy_ranks = pd.concat([accuracy_ranks, Accuracy_rankMean.to_frame().T])
print('\n\n ACCURACY RANK TABLE')
print('\n', accuracy_ranks)

#F1 Score Rank Table
f1_ranks = F1Scores_df[model_cols].rank(axis=1, ascending=False)
f1_ranks.insert(0, 'Fold', range(1,11)) #adding the fold column
F1_rankMean = f1_ranks.mean().rename('Mean')
F1_rankMean['Fold'] = '_'
f1_ranks = pd.concat([f1_ranks, F1_rankMean.to_frame().T])
print('\n\n F1SCORE RANK TABLE')
print('\n', f1_ranks)

#Training Time Rank Table
time_ranks = trainingTime_df[model_cols].rank(axis=1, ascending=True)
time_ranks.insert(0, 'Fold', range(1,11)) #adding the fold column
time_rankMean = time_ranks.mean().rename('Mean')
time_rankMean['Fold'] = '_'
time_ranks = pd.concat([time_ranks, time_rankMean.to_frame().T])
print('\n\n TRAINING TIME RANK TABLE')
print('\n', time_ranks)
```

### The Friedman Test for Accuracy 

```python
#Friedman Numerator = n * (Rank_meanLR -2)^2 + (Rank_meanRF -2)^2 + (Rank_meanSVM)^2
#Denominator = (sum of Individual deviation from the R_bar)^2 / n * (k-1)

#I'll use a table to calculate the individual deviation
print(f"The average for the Accuracy of each algorithm: {Accuracy_rankMean}")
```

```python
#Caluclating the Numerator
n = 10
k = 3
R_bar = 2
Acc_numerator = (((Accuracy_rankMean['LR'] - R_bar)**2) + ((Accuracy_rankMean['RF']-R_bar)**2) + ((Accuracy_rankMean['SVM']-R_bar)**2)) * n
print('Accuracy Numerator = ', Acc_numerator)
```

```python
#Calculating the Denominator - Accuracy
RmeanDev_table = (accuracy_ranks[model_cols] - R_bar)**2 #Calculating R - R_BAR for each cell
RmeanDev_table.drop(index= 'Mean', inplace=True, axis=0)
RmeanDev_table.insert(0, 'Fold', range(1,11)) #adding the fold column
Rmean_sum = RmeanDev_table.sum().rename('Sum')
Rmean_sum['Fold'] = '_'
RmeanDev_table = pd.concat([RmeanDev_table, Rmean_sum.to_frame().T])
print('\n\n Accuracy Squared Deviation Table')
print('\n', RmeanDev_table)

Acc_denominator = (Rmean_sum['LR'] + Rmean_sum['RF'] + Rmean_sum['SVM']) / (n *(k-1))
print('\n Accuracy Denominator = ', Acc_denominator)
```

```python
#The final equation - For Accuracy
Acc_Fried = Acc_numerator / Acc_denominator
print(f"Friedman Statistic for Accuracy: {Acc_Fried:.2f}")
```

### The Friedman Test for F1 Score 

```python
print(f"The average score for the F1 Measure of each algorithm: {F1_rankMean}")
```

```python
#Calculating the Numerator - F1 Score
f1_numerator = n * (((F1_rankMean['LR'] - R_bar)**2) + ((F1_rankMean['RF']-R_bar)**2) + ((F1_rankMean['SVM']-R_bar)**2))
print(f"F1 Numerator: {f1_numerator:.2f}")
```

```python
#Calculating the Denominator - F1 Score
f1_Rbar_Dev = (f1_ranks[model_cols] - R_bar)**2 #Calculating R - R_BAR for each cell
f1_Rbar_Dev.drop(index= 'Mean', inplace=True, axis=0) #dropping the row of means
f1_Rbar_Dev.insert(0, 'Fold', range(1,11)) #adding the fold column
f1_Rbar_sum = f1_Rbar_Dev.sum().rename('Sum')
f1_Rbar_sum['Fold'] = '_'
f1_Rbar_Dev = pd.concat([f1_Rbar_Dev, f1_Rbar_sum.to_frame().T])
print('\n\n F1 Score Squared Deviation Table')
print('\n', f1_Rbar_Dev)

f1_denominator = (f1_Rbar_sum['LR'] + f1_Rbar_sum['RF'] + f1_Rbar_sum['SVM']) / (n *(k-1))
print(f"\nF1 Denominator: {f1_denominator}")
```

```python
#The final equation - F1 Score
f1_Fried = f1_numerator / f1_denominator
print(f"Friedman Statistic for F1 Measure: {f1_Fried:.2f}")
```

### Friedman Test for Training Time

```python
print(f"The average score for the Training time of each algorithm: {time_rankMean}")
```

```python
#Calculating the Numerator - Training time
trainTime_numerator = n * (((time_rankMean['LR'] - R_bar)**2) + ((time_rankMean['RF']-R_bar)**2) + ((time_rankMean['SVM']-R_bar)**2))
print(f"Time Numerator: {trainTime_numerator}")
```

```python
#Calculating the Denominator - Training Time
time_Rbar_Dev = (time_ranks[model_cols] - R_bar)**2 #Calculating R - R_BAR for each cell
time_Rbar_Dev.drop(index= 'Mean', inplace=True, axis=0) #drop the row of means
time_Rbar_Dev.insert(0, 'Fold', range(1,11)) #adding the fold column
time_Rbar_sum = time_Rbar_Dev.sum().rename('Sum')
time_Rbar_sum['Fold'] = '_'
time_Rbar_Dev = pd.concat([time_Rbar_Dev, time_Rbar_sum.to_frame().T])
print('\n\n Training Time Squared Deviation Table')
print('\n', time_Rbar_Dev)

time_denominator = (time_Rbar_sum['LR'] + time_Rbar_sum['RF'] + time_Rbar_sum['SVM']) / (n *(k-1))
print(f"\nF1 Time Denominator: {time_denominator}")
```

```python
#The final equation - Training Time
time_Fried = trainTime_numerator / time_denominator
print(f"Friedman Statistic for Training time is: {time_Fried:.2f}")
```

# Nemenyi Test

```python
#Criticial Difference is given as CD = q * SQRT[k(k+1)/6n]
#Where q is the q-value at alpha α = 0.05 and k = 3 which is 2.343

q = 2.343   # for alpha=0.05, k=3
cd = q * (np.sqrt(k * (k + 1) / (6 * n)))

print(f'Critical Difference(CD) = {cd:.2f}')
```
